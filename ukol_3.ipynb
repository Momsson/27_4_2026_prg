# Inicializacia

```python
import numpy as np
import matplotlib.pyplot as plt


d = np.load("mpl_data.npz")
tf_test = d["tf_test"]
```

# tf_test.shape == (30, 60)

# 1. Průměrná TF v čase (signál)

```python

mean_tf = np.mean(tf_test, axis=0)

plt.figure()
plt.plot(mean_tf)
plt.xlabel("Minuty")
plt.ylabel("Tepová frekvence (BPM)")
plt.title("Průměrná tepová frekvence v čase")
plt.show()
```

# 2. Boxplot – 5 fází

```python
faze_data = [
    tf_test[:, 0:10].ravel(),
    tf_test[:, 10:20].ravel(),
    tf_test[:, 20:40].ravel(),
    tf_test[:, 40:50].ravel(),
    tf_test[:, 50:60].ravel()
]

faze_labels = ["Klid", "Zahřívání", "Aerobní", "Sprint", "Regenerace"]

plt.figure()
plt.boxplot(faze_data, tick_labels=faze_labels)  #
plt.ylabel("Tepová frekvence (BPM)")
plt.title("Rozdělení TF podle fází zátěže")
plt.grid(True)
plt.show()
```

# 3. Histogram – klid vs sprint

```python
klid = tf_test[:, 0:10].ravel()
sprint = tf_test[:, 40:50].ravel()

plt.figure()
plt.hist(klid, alpha=0.5, label="Klid")
plt.hist(sprint, alpha=0.5, label="Sprint")
plt.xlabel("Tepová frekvence (BPM)")
plt.ylabel("Počet")
plt.title("Porovnání TF: Klid vs Sprint")
plt.legend()
plt.grid(True)
plt.show()
```
